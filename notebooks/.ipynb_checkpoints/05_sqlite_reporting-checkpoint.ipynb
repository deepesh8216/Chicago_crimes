```python
import sqlite3
import pandas as pd
from pathlib import Path

# Connect to SQLite database
DB_PATH = Path("../data/database/chicago_crime.db")

connection = sqlite3.connect(DB_PATH)

# Check total crime records
total_records = pd.read_sql_query(
    "SELECT COUNT(*) AS total_records FROM crimes",
    connection
)

display(total_records)

# Display first 10 records
crime_preview = pd.read_sql_query(
    "SELECT * FROM crimes LIMIT 10",
    connection
)

display(crime_preview)

# Display database table names
tables = pd.read_sql_query(
    """
    SELECT name
    FROM sqlite_master
    WHERE type = 'table'
    """,
    connection
)

display(tables)
```

```python
# Yearly crime count
yearly_crimes = pd.read_sql_query(
    """
    SELECT
        year,
        COUNT(*) AS total_crimes
    FROM crimes
    GROUP BY year
    ORDER BY year
    """,
    connection
)

display(yearly_crimes)
```

```python
# Top 5 crime categories
top_crimes = pd.read_sql_query(
    """
    SELECT
        primary_type,
        COUNT(*) AS crime_count,
        ROUND(
            COUNT(*) * 100.0 /
            (SELECT COUNT(*) FROM crimes),
            2
        ) AS percentage
    FROM crimes
    GROUP BY primary_type
    ORDER BY crime_count DESC
    LIMIT 5
    """,
    connection
)

display(top_crimes)
```

```python
# Arrest statistics
arrest_summary = pd.read_sql_query(
    """
    SELECT
        COUNT(*) AS total_crimes,
        SUM(arrest) AS arrests,
        COUNT(*) - SUM(arrest) AS no_arrests,
        ROUND(AVG(arrest) * 100, 2) AS arrest_percentage
    FROM crimes
    """,
    connection
)

display(arrest_summary)
```

```python
connection.close()
print("SQLite connection closed.")
```

```python
import sqlite3
import pandas as pd
from pathlib import Path

# Reconnect to SQLite database
DB_PATH = Path("../data/database/chicago_crime.db")

connection = sqlite3.connect(DB_PATH)

cursor = connection.cursor()

print("SQLite database reconnected successfully.")
```

```python
cursor.execute("""
    DROP VIEW IF EXISTS vw_crime_yearly
""")

cursor.execute("""
    CREATE VIEW vw_crime_yearly AS
    SELECT
        year,
        COUNT(*) AS total_crimes,
        SUM(arrest) AS total_arrests,
        ROUND(AVG(arrest) * 100, 2) AS arrest_percentage
    FROM crimes
    GROUP BY year
""")

connection.commit()

print("vw_crime_yearly created successfully.")
```

```python
cursor.execute("""
    DROP VIEW IF EXISTS vw_crime_by_category
""")

cursor.execute("""
    CREATE VIEW vw_crime_by_category AS
    SELECT
        primary_type,
        COUNT(*) AS crime_count,
        ROUND(
            COUNT(*) * 100.0 /
            (SELECT COUNT(*) FROM crimes),
            2
        ) AS percentage,
        SUM(arrest) AS total_arrests
    FROM crimes
    GROUP BY primary_type
""")

connection.commit()

print("vw_crime_by_category created successfully.")
```

```python
# Verify yearly view
display(pd.read_sql_query(
    "SELECT * FROM vw_crime_yearly ORDER BY year",
    connection
))

# Verify category view
display(pd.read_sql_query(
    """
    SELECT *
    FROM vw_crime_by_category
    ORDER BY crime_count DESC
    LIMIT 10
    """,
    connection
))
```

```python
views = pd.read_sql_query(
    """
    SELECT name
    FROM sqlite_master
    WHERE type = 'view'
    ORDER BY name
    """,
    connection
)

display(views)
```

```python
import sqlite3
import pandas as pd
from pathlib import Path

BASE_DIR = Path("..").resolve()
DB_PATH = BASE_DIR / "data" / "database" / "chicago_crime.db"

connection = sqlite3.connect(DB_PATH)

relationship_queries = {
    "District": """
        SELECT COUNT(*) AS matched_records
        FROM crimes c
        INNER JOIN dim_district d
        ON c.district_code = d.district_code
    """,

    "Ward": """
        SELECT COUNT(*) AS matched_records
        FROM crimes c
        INNER JOIN dim_ward w
        ON c.ward_no = w.ward
    """,

    "IUCR": """
        SELECT COUNT(*) AS matched_records
        FROM crimes c
        INNER JOIN dim_iucr i
        ON TRIM(c.iucr_code) = TRIM(i.iucr)
    """,

    "Community": """
        SELECT COUNT(*) AS matched_records
        FROM crimes c
        INNER JOIN dim_community cm
        ON c.community_code = cm.community_code
    """,

    "Police Beat": """
        SELECT COUNT(*) AS matched_records
        FROM crimes c
        INNER JOIN dim_police_beat pb
        ON c.beat_num = pb.beat_num
    """
}

print("DIMENSION RELATIONSHIP CHECK")
print("-" * 45)

for name, query in relationship_queries.items():

    result = pd.read_sql_query(query, connection)

    print(
        f"{name}: "
        f"{result.iloc[0]['matched_records']} matched crime records"
    )

connection.close()
```
